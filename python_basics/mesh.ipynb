```python
#test
import numpy as np
import matplotlib.pyplot as plt

# Axe horizontal x : 2 valeurs → 2 colonnes
x = np.array([10, 20])

# Axe vertical y : 3 valeurs → 3 lignes
y = np.array([1, 2, 3])

XX, YY = np.meshgrid(x, y)

print("XX :")
print(XX)

print("\nYY :")
print(YY)

print("\nDimensions de XX :", XX.shape)
print("Dimensions de YY :", YY.shape)
```

```python
Z = np.zeros((y.size, x.size))

for i, valeur_x in enumerate(x):
    for j, valeur_y in enumerate(y):
        Z[j, i] = valeur_x + 100 * valeur_y

print(Z)
print("Dimensions de Z :", Z.shape)
```

```python
fig, ax = plt.subplots(figsize=(7, 5))

graphique = ax.pcolormesh(
    XX,
    YY,
    Z,
    shading="nearest",
    cmap="viridis"
)

# Écrire les valeurs dans les cases
for j in range(y.size):          # lignes
    for i in range(x.size):      # colonnes
        ax.text(
            XX[j, i],
            YY[j, i],
            f"{Z[j, i]:.0f}",
            ha="center",
            va="center",
            color="white",
            fontsize=12
        )

ax.set_xlabel("x — colonnes")
ax.set_ylabel("y — lignes")
ax.set_title("Z[ligne, colonne] = Z[j, i]")

ax.set_xticks(x)
ax.set_yticks(y)

plt.colorbar(graphique, ax=ax, label="Valeur de Z")
plt.show()
```
