## Random Datasets

### Genearte a list of 1000 random floats within 0 and 1.

```python
import random

valeurs = []

for i in range(1000):
    valeurs.append(random.random())

print(valeurs)

print('---')

# autre méthode

valeur = []
valeurs = [random.random() for i in range(1000)]

print(valeurs)

import matplotlib.pyplot as plt

plt.hist(valeurs, bins=100, edgecolor="black")
plt.xlabel("Valeurs")
plt.ylabel("Fréquence")
plt.title("100 valeurs suivant un pur random")
plt.show()
```

### Generate a list of 1000 random integers within 0 and 1.

```python
import random
import matplotlib.pyplot as plt

valeurs = [random.randint(1, 5) for _ in range(1000)] -- randint

frequences = [valeurs.count(x) for x in range(1, 6)]

plt.bar(
    range(1, 6),
    frequences,
    edgecolor="black"
)

plt.xlabel("Valeurs")
plt.ylabel("Fréquence")
plt.title("Distribution de 1000 entiers aléatoires")
plt.xticks(range(1, 6))

plt.show()
```

### Generate a list of 1000 numbers following normal distribution with 10 as mean, and 1 as standard derivation.

```python
valeurs = [
    round(random.normalvariate(10, 2))
    for _ in range(1000)
]

entiers = range(min(valeurs), max(valeurs) + 1)
frequences = [valeurs.count(x) for x in entiers]

plt.bar(
    entiers,
    frequences,
    edgecolor="black"
)

plt.xlabel("Valeurs entières")
plt.ylabel("Fréquence")
plt.title("Distribution normale arrondie")
plt.show()
```
